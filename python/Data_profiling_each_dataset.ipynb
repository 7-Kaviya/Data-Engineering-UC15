## Visualizations

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11
```

```python
dataset_counts = {
    "Sales": len(sales),
    "Asset": len(asset),
    "Customer Master": len(customer),
    "Procurement": len(procurement),
    "Accounts": len(account)
}

plt.figure()
plt.bar(dataset_counts.keys(), dataset_counts.values())
plt.title("ERP Dataset Record Count")
plt.xlabel("Dataset")
plt.ylabel("Number of Records")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
category_col = None

for col in sales.columns:
    if col.lower().replace("_", " ") in [
        "product category",
        "category"
    ]:
        category_col = col
        break

if category_col:
    category_counts = sales[category_col].value_counts()

    plt.figure()
    plt.bar(category_counts.index, category_counts.values)
    plt.title("Sales by Product Category")
    plt.xlabel("Product Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("Product category column not found.")
```

```python
condition_col = None
for col in asset.columns:
    if "condition" in col.lower():
        condition_col = col
        break

if condition_col:

    plt.figure()
    plt.hist(
        pd.to_numeric(
            asset[condition_col],
            errors="coerce"
        ).dropna(),
        bins=20
    )

    plt.title("Asset Condition Score Distribution")
    plt.xlabel("Condition Score")
    plt.ylabel("Number of Assets")
    plt.tight_layout()
    plt.show()
else:
    print("Condition score column not found.")
```

```python
customer_type_col = None
for col in customer.columns:
    if "customer_type" in col.lower() or "customer type" in col.lower():
        customer_type_col = col
        break

if customer_type_col:
    type_counts = customer[customer_type_col].value_counts()
    plt.figure()
    plt.bar(
        type_counts.index.astype(str),
        type_counts.values
    )

    plt.title("Customer Distribution by Customer Type")
    plt.xlabel("Customer Type")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```

```python
proc_amount_col = None
for col in procurement.columns:
    col_clean = col.lower().replace("_", " ")
    if col_clean in [
        "total amount",
        "procurement amount",
        "amount",
        "total cost",
        "purchase amount"
    ]:
        proc_amount_col = col
        break

if proc_amount_col:
    values = pd.to_numeric(
        procurement[proc_amount_col],
        errors="coerce"
    ).dropna()

    plt.figure()
    plt.hist(values, bins=30)

    plt.title("Procurement Amount Distribution")
    plt.xlabel("Procurement Amount")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
else:
    print("Procurement amount column not found.")
```

```python
account_amount_col = None
for col in account.columns:
    col_clean = col.lower().replace("_", " ")
    if col_clean in [
        "amount",
        "transaction amount",
        "debit",
        "credit",
        "balance",
        "total amount"
    ]:
        account_amount_col = col
        break

if account_amount_col:
    values = pd.to_numeric(
        account[account_amount_col],
        errors="coerce"
    ).dropna()
    plt.figure()
    plt.hist(values, bins=30)

    plt.title("Accounts Financial Amount Distribution")
    plt.xlabel("Amount")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
else:
    print("Financial amount column not found.")
```
