```python
print("\n===== MISSING VALUES =====")

for name, df in {
    "Finance": finance,
    "Employee": employee,
    "Inventory": inventory,
    "Manufacturing": manufacturing,
    "Warehouse": warehouse
}.items():

    print(f"\n{name}:")
    print(df.isnull().sum())

    print("Total missing values:", df.isnull().sum().sum())
```

```python
print("\n===== DUPLICATE ROWS =====")

for name, df in {
    "Finance": finance,
    "Employee": employee,
    "Inventory": inventory,
    "Manufacturing": manufacturing,
    "Warehouse": warehouse
}.items():

    print(f"{name}: {df.duplicated().sum()}")
```

```python
print("\n===== DESCRIPTIVE STATISTICS =====")

print("\nFinance:")
print(finance.describe(include="all"))

print("\nEmployee:")
print(employee.describe(include="all"))

print("\nInventory:")
print(inventory.describe(include="all"))

print("\nManufacturing:")
print(manufacturing.describe(include="all"))

print("\nWarehouse:")
print(warehouse.describe(include="all"))
```

```python
print("\n===== NUMERICAL AND CATEGORICAL COLUMNS =====")

datasets = {
    "Finance": finance,
    "Employee": employee,
    "Inventory": inventory,
    "Manufacturing": manufacturing,
    "Warehouse": warehouse
}

for name, df in datasets.items():

    numerical_columns = df.select_dtypes(
        include=["int64", "float64"]
    ).columns

    categorical_columns = df.select_dtypes(
        include=["object", "string", "category"]
    ).columns

    print(f"\n{name}")

    print("Numerical Columns:")
    print(list(numerical_columns))

    print("Categorical Columns:")
    print(list(categorical_columns))
```

```python
print("\n===== CATEGORICAL VALUE COUNTS =====")

for name, df in datasets.items():

    print(f"\n================ {name} ================")

    categorical_columns = df.select_dtypes(
        include=["object", "string", "category"]
    ).columns

    for column in categorical_columns:

        print(f"\n{column}:")
        print(df[column].value_counts(dropna=False).head(20))
```
