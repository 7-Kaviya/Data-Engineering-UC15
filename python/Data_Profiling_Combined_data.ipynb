## Visualizations

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 11
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=data['total_amount'].dropna())

plt.title("Boxplot of Sales Amount")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
plt.figure(figsize=(10,5))

data['product_name'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Products by Number of Sales")
plt.xlabel("Product Name")
plt.ylabel("Number of Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data['total_amount'].dropna(), kde=True)

plt.title("Distribution of Sales Amount")
plt.xlabel("Total Sales Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

data['sales_region'].value_counts().plot(kind='bar')

plt.title("Sales by Region")
plt.xlabel("Sales Region")
plt.ylabel("Number of Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7,5))

data['customer_type'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title("Customer Type Distribution")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=data['condition_score'].dropna())

plt.title("Boxplot of Asset Condition Score")
plt.ylabel("Condition Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data['total_cost'].dropna(), kde=True)

plt.title("Distribution of Procurement Cost")
plt.xlabel("Total Procurement Cost")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=data['total_cost'].dropna())

plt.title("Boxplot of Procurement Cost")
plt.ylabel("Total Procurement Cost")
plt.show()
```

```python
plt.figure(figsize=(8,5))

data['transaction_type'].value_counts().plot(kind='bar')

plt.title("Financial Transaction Type Distribution")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
debit = data['debit_amount'].fillna(0).sum()
credit = data['credit_amount'].fillna(0).sum()

plt.figure(figsize=(6,5))

plt.bar(['Debit', 'Credit'], [debit, credit])

plt.title("Total Debit vs Credit Amount")
plt.ylabel("Amount")
plt.show()
```

```python
missing = data.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(12,6))

missing.plot(kind='bar')

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=90)
plt.show()
```
